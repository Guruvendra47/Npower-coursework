```sql
CREATE CATALOG IF NOT EXISTS CTG_SQL_PRACTICE;
USE CATALOG CTG_SQL_PRACTICE;
CREATE SCHEMA IF NOT EXISTS SM_SQL_PRACTICE;
USE SCHEMA SM_SQL_PRACTICE;



-- Load tables from GitHub-linked workspace folder
-- Source CSVs stay in your GitHub repo; read_files reads them directly
CREATE OR REPLACE TABLE departments AS
SELECT
  department_id,
  department_name,
  location
FROM read_files('/Workspace/Users/guruvendra47@gmail.com/Npower-coursework/SQL/departments.csv', format => 'csv', header => true);

CREATE OR REPLACE TABLE employees AS
SELECT
  employee_id,
  name,
  department_id,
  CAST(manager_id AS INT) AS manager_id,
  salary,
  hire_date
FROM read_files('/Workspace/Users/guruvendra47@gmail.com/Npower-coursework/SQL/employees.csv', format => 'csv', header => true);

CREATE OR REPLACE TABLE projects AS
SELECT
  project_id,
  project_name,
  department_id,
  budget
FROM read_files('/Workspace/Users/guruvendra47@gmail.com/Npower-coursework/SQL/projects.csv', format => 'csv', header => true);

CREATE OR REPLACE TABLE employee_projects AS
SELECT
  employee_id,
  project_id,
  hours_worked
FROM read_files('/Workspace/Users/guruvendra47@gmail.com/Npower-coursework/SQL/employee_projects.csv', format => 'csv', header => true);

-- Q1 Common table Expression(CTE)

with singh as (
SELECT
  e.employee_id,
  e.name,
  d.department_name,
  e.manager_id,
  e.salary,
  e.hire_date,
  dense_rank() OVER(ORDER BY e.salary DESC) AS ranked
FROM employees e
left join departments d
ON e.department_id = d.department_id
)

SELECT *
FROM singh
WHERE ranked = 2;

--- Q2 Self joined check respective 

SELECT 
  e.employee_id,
  e.name as employees_name,
  m.employee_id as manager_id,
  m.name as manager_name
FROM employees e
LEFT JOIN employees m
ON m.employee_id = e.manager_id;

--Q3 Comparision between employee and Manager

with singh as (
SELECT
 e.employee_id,
  e.name as employees_name,
  e.salary as employees_salary,
  m.employee_id as manager_id,
  m.name as manager_name,
  m.salary as manager_salary
FROM employees e
LEFT JOIN employees m
ON m.employee_id = e.manager_id
)

SELECT *
FROM singh
WHERE manager_salary < employees_salary;








```
