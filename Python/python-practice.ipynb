```python
employees = [
    {"name": "Bob", "manager": "Alice"},
    {"name": "Charlie", "manager": "Alice"},
    {"name": "David", "manager": "Eva"},
    {"name": "Alice", "manager": None},
    {"name": "Eva", "manager": None},
]
def count_direct_reports(employees):
    dic = {}
    for i in employees:
        if i["manager"] != None:
            dic[i["manager"]] = dic.get(i["manager"], 0) + 1
    return dic

g = count_direct_reports(employees)
print(g) 
```

Q1 “You’re given a list of integers. Write a function find_missing_number(nums) that finds the one missing number from a sequence of consecutive integers starting at 1. The list has n numbers but is missing exactly one number from the range 1 to n+1.”

Example: find_missing_number([1, 2, 4, 5]) → 3 is missing (the full sequence should be 1,2,3,4,5, but 3 isn’t there)

```python
def find_missing_number(nums):
    for i in range(1,len(nums)+1):
        if i not in nums:
            return i 
        
g = find_missing_number([1, 2, 4, 5])
print(g)
```
