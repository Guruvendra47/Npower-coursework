````python
# checking whitespaces before and after

print("before")
for i in df.columns:
    if df[i].dtype == "object":
        whitespaces = (df[i].notna() & (df[i] != df[i].str.strip()))
        count_whitespaces = whitespaces.sum()
        if count_whitespaces > 0:
            print(f"{i}: {count_whitespaces}")
        else:
            print(f"{i}: {count_whitespaces}")

print("removing whitespaces....")
for i in df.columns:
    if df[i].dtype == "object":
        df[i] = df[i].str.strip()



print("after removing")
for i in df.columns:
    if df[i].dtype == "object":
        whitespaces = (df[i].notna() & (df[i] != df[i].str.strip()))
        count_whitespaces = whitespaces.sum()
        if count_whitespaces > 0:
            print(f"{i}: {count_whitespaces}")
        else:
            print(f"{i}: {count_whitespaces}")

# whitespaces are done then we move to checking data type 
            
````

### Changing Data Types

1. Identify columns with incorrect data types.
2. Convert to appropriate types: `object`, `int64`, `float64`, or `datetime64[ns]`.
3. **Finding:** `LaunchDate` and `DecayDate` are currently stored as `object` and must be converted to `datetime64[ns]`.

#### Note:
- Use `pd.to_numeric()` for int/float conversions and `pd.to_datetime()` for date conversions — especially when data may contain invalid or dirty values.
- Use `.astype()` only as a final clean cast when the data is already numeric.

#### Examples:
```python
# Recommended — handles dirty data gracefully
df["LaunchDate"] = pd.to_datetime(df["LaunchDate"], format="mixed", errors="coerce")

# Not recommended — fails on dirty data
df["LaunchDate"] = df["LaunchDate"].astype("datetime64[ns]")

# Recommended — coerces invalid values to NaN before casting
df["age"] = pd.to_numeric(df["age"], errors="coerce").astype("Int64")

# Not recommended — fails on non-numeric values
df["age"] = df["age"].astype("Int64")
```

````python
df["signup_date"] = pd.to_datetime(df["signup_date"],format="mixed", errors="coerce")

df["age"] = pd.to_numeric(df["age"], errors="coerce").astype("Int64")

df.info()
````

````python
# check null values 

cnt = df.isnull().sum()
percentage = (cnt/len(df)*100).round(2)
dic = {"no of missing values": cnt, "percentage": percentage}
df1 = pd.DataFrame(dic)
print(df1)
````

````python
# name has 1 missing value
# age has 3 missing value
# purchase amount 4 missing value

# remove the na name row
print(df[df["name"].isnull()])
# we remove it 
df = df.dropna(subset=["name"])

# for age we can add average value
avg_value = df["age"].mean().round(0)

df["age"] = df["age"].fillna(avg_value)
df.isnull().sum()

# purchase amount also i add avg purchase amount

avg_value = int(df["purchase_amount"].mean())

df["purchase_amount"] = df["purchase_amount"].fillna(avg_value)

df.isnull().sum()
````
