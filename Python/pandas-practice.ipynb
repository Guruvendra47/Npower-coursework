```python
employees = [
    {"name": "Bob", "manager": "Alice"},
    {"name": "Charlie", "manager": "Alice"},
    {"name": "David", "manager": "Eva"},
    {"name": "Alice", "manager": None},
    {"name": "Eva", "manager": None},
]
def count_direct_reports(employees):
    dic = {}
    for i in employees:
        if i["manager"] != None:
            dic[i["manager"]] = dic.get(i["manager"], 0) + 1
    return dic

g = count_direct_reports(employees)
print(g) 
```

```python
employees = [
    {"name": "Bob", "manager": "Alice"},
    {"name": "Charlie", "manager": "Alice"},
    {"name": "David", "manager": "Eva"},
    {"name": "Alice", "manager": None},
    {"name": "Eva", "manager": None},
]
def count_direct_reports(employees):
    dic = {}
    for i in employees:
        if i["manager"] != None:
            if i["manager"] in dic:
                dic[i["manager"]] += 1
            else:
                 dic[i["manager"]] = 1
    return dic

g = count_direct_reports(employees)
print(g) 
```
